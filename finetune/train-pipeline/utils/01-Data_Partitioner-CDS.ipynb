#### Data spliting generator

```python
import json, os

def train_test_file_split(train_range, test_range, dev_range, save_path, is_list=False):
    # internal utils
    def generate_file_list(start, end):
        return [f"bacteria_{i}.csv" for i in range(start, end+1)]
    
    # generate
    if not is_list:
        train_files = generate_file_list(*train_range)
        test_files  = generate_file_list(*test_range)
        dev_files   = generate_file_list(*dev_range)
    else:
        train_files = train_range
        test_files  = test_range
        dev_files   = dev_range

    data = {
        "train": train_files,
        "test":  test_files,
        "dev":   dev_files,
    }

    os.makedirs(os.path.dirname(save_path), exist_ok=True)

    with open(save_path, "w") as json_file:
        json.dump(data, json_file, indent=4)
    
    return True
    
```

```python
train_ratio = 0.7
dev_ratio   = 0.2
test_ratio  = 0.1

random.shuffle(files)
train_size = int(len(files) * train_ratio)
dev_size   = int(len(files) * dev_ratio)

train_files = files[:train_size]
dev_files   = files[train_size:train_size + dev_size]
test_files  = files[train_size + dev_size:]

train_test_file_split(train_files, test_files, dev_files, "../data/cds/BacteriaCDS-ORFSeq2Class-v2/data_index.json", is_list=True)
```
